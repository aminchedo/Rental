## Inspect the Data

```python
with open('../data/processed/train.json', 'r') as f:
    train_data = json.load(f)

with open('../data/processed/test.json', 'r') as f:
    test_data = json.load(f)

print(f"Training samples: {len(train_data['labels'])}")
print(f"Test samples: {len(test_data['labels'])}")
print(f"Features per sample: {len(train_data['features'][0])}")
```

## Visualize Data Distribution

```python
import matplotlib.pyplot as plt

X_train = np.array(train_data['features'])
y_train = np.array(train_data['labels'])

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.scatter(X_train[:, 0], X_train[:, 1], c=y_train, cmap='viridis', alpha=0.5)
plt.xlabel('Feature 0')
plt.ylabel('Feature 1')
plt.title('First Two Features')
plt.colorbar(label='Label')

plt.subplot(1, 2, 2)
plt.hist(y_train, bins=2, edgecolor='black')
plt.xlabel('Label')
plt.ylabel('Count')
plt.title('Label Distribution')

plt.tight_layout()
plt.show()
```
